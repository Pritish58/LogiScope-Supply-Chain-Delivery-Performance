```python
# CELL 1: Imports and load data

import pandas as pd
import numpy as np
from pathlib import Path

DATA_PATH = Path("../../APL_Logistics(1).csv")

if not DATA_PATH.exists():
    DATA_PATH = Path("../data/raw/APL_Logistics.csv")

df = pd.read_csv(DATA_PATH, encoding="latin1")

print("Original shape:", df.shape)
```

```python
# CELL 2: Remove duplicate records

df = df.drop_duplicates()

print("Shape after duplicate removal:", df.shape)
```

```python
# CELL 3: Handle missing values

numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(include="object").columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")

print("Missing values handled.")
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
# CELL 4: Standardize column names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace(r"[^a-z0-9_]", "", regex=True)
)

display(df.head())
```

```python
# CELL 5: Validate shipping duration

df["days_for_shipping_real"] = pd.to_numeric(
    df["days_for_shipping_real"], errors="coerce"
)

df["days_for_shipment_scheduled"] = pd.to_numeric(
    df["days_for_shipment_scheduled"], errors="coerce"
)

df = df.dropna(subset=[
    "days_for_shipping_real",
    "days_for_shipment_scheduled"
])

df = df[
    (df["days_for_shipping_real"] >= 0) &
    (df["days_for_shipment_scheduled"] >= 0)
]

print("Validated shape:", df.shape)
```

```python
# CELL 6: Save cleaned dataset

OUTPUT_DIR = Path("../data/processed")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df.to_csv(OUTPUT_DIR / "logistics_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
